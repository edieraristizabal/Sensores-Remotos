## Autenticación

Primero debemos autenticarnos en el CDSE.

```python
connection.authenticate_oidc()
```

Comienza con un cubo de datos simple: una pequeña porción espacio-temporal de datos de `SENTINEL2_L2A`:

```python
cube = connection.load_collection(
    "SENTINEL2_L2A",
    bands=["B04", "B03", "B02", "B08", "SCL"],
    temporal_extent=("2025-07-01", "2025-10-31"),
    spatial_extent={
        "west": -75,
        "south": 5,
        "east": -74,
        "north": 6,
        "crs": "EPSG:4326",
    },
    max_cloud_cover=50,
)
```

## Máscara para nubes

En general, para hacer que los datos crudos sean más útiles, normalmente queremos eliminar los píxeles de nubes y trabajar solo con datos sin nubes. Es muy común que los datos de observación de la Tierra tengan capas de enmascaramiento separadas que, por ejemplo, indican si un píxel está cubierto por un (tipo de) nube o no. Para Sentinel-2, una de esas capas es la capa de "clasificación de escena" (nombre de banda "SCL") generada por el algoritmo Sen2Cor.

Con openEO y la biblioteca cliente de openEO para Python, podemos tomar esta banda "SCL" (que ya incluimos antes en la llamada a `load_collection`) y aplicar el enmascaramiento de nubes de la siguiente manera.

Primero construimos una máscara de nubes binaria a partir de los valores 3 (sombras de nubes), 8 (probabilidad media de nubes) y 9 (probabilidad alta de nubes) de la SCL:

```python
# Aislamos la banda SCL
scl_band = cube.band("SCL")

# Creamos una máscara booleana: True donde SCL es 4 (veg) O 5 (suelo)
cloud_mask = (scl_band == 3) | (scl_band == 8) | (scl_band == 9)
```

Aplica la máscara de nubes, y construye el compuesto de nuevo:

```python
cube_masked = cube.mask(cloud_mask)
```

## Cálculo del NDVI

```python
# Para crear un cubo con un NDVI por cada imagen válida.
ndvi_cube = cube_masked.ndvi(nir="B08", red="B04")
```

Podemos crear un "compuesto" (un solo mapa) tomando el MÁXIMO NDVI de cada píxel a lo largo del mes. Esto ayuda a obtener la vegetación más saludable
y reduce aún más los efectos residuales de nubes/bruma.

```python
ndvi_composite = ndvi_cube.reduce_dimension(dimension="t", reducer="max")
```

Todo lo anterior solo ha definido el flujo de trabajo. Ahora, enviamos el trabajo al backend de CDSE para que se procese y descargamos el resultado como un GeoTIFF.

```python
ndvi_composite.download("ndvi.tif")
```
